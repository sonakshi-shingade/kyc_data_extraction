## 1. KYC Data Extraction System (Images) 

- Use Case: Finance / HR / Onboarding / Banking
  
- Problem Statement: **Users upload scanned images of PAN or Aadhaar cards. The system 
extracts data like Name, DOB, PAN/Aadhaar Number using OCR (e.g., Pytesseract) and 
stores the extracted data in the database.**

 
- Key Fields to Extract/Store: Name, DOB, PAN/Aadhaar Number, Gender, Father’s Name 
- Core APIs: 
- /register, /login, /upload-kyc, /kyc-records/<username>, /kyc-details/<doc_id>
